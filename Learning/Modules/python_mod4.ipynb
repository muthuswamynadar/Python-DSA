```python
def sum(a):
    n=[]
    for i in a:
        if type(i)==int or type(i)==float:
          n.append(i)
    return n
```

```python
def sum1(*args):
    s=0
    for i in args:
        s=s+i
    return s

```

```python
sum1(7,100,200)
```

```python
def min(*args):
    n=[]
    for i in args:
        n.append(i)
    return n
```

```python
min(1,2,3,4,5,6,7,7)
```

# lambda

```python
square=lambda m: m*m
print(square(9))
```
