```python
# questions : Write the functions takes the list as input
def onlynum(a):
    n=[]
    for i in a:
        if type(i)==int or type(i)==float:
            n.append(i)
    return n
```

```python
onlynum([10,20,30,40,"muthu","money",70])
```

```python
# question 2 
# having the nested list on the crew

def next(a):
    n=[]
    for i in a:
        if type(i)==list:
            for j in i:
                if type(j)==int or type(j)==float:
                    n.append(j)
        else:
            if type(i)==int or type(i)==float:
                    n.append(i)
                    
    return n

```

```python
next([1,2,3,[2,2],"Swamy"])
```
