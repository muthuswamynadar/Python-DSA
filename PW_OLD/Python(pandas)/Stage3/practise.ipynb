# Data Visualization

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,20,15,17,13]
plt.plot(x,y,label="lineplot")
plt.xlabel("Height")
plt.ylabel("width")
plt.title("Swamy plot")
plt.legend()
plt.show()
```

```python
plt.bar(x,y,label="X-one",color='orange')

plt.legend()
plt.show()
```

```python
plt.scatter(x,y,label="sales",color='green')
plt.title("Main")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import seaborn as sns
sns.histplot(x,kde=False)
plt.title("sam")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
dataz= pd.DataFrame({
    "score": [45, 67, 78, 90, 82, 56, 75, 80, 60, 85],
    "category": ["A", "B", "A", "B", "A", "B", "A", "B", "A", "B"]
})
```

```python
sns.histplot(dataz["score"], kde=True)
plt.title("Seaborn Histogram with KDE")
plt.show()
```

```python
sns.boxplot(x="category", y="score", data=dataz)
plt.title("Seaborn Boxplot by Category")
plt.show()
```

```python
correlation = dataz.select_dtypes(include='number').corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Seaborn Heatmap")
plt.show()
```
